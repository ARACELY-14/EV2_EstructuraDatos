```python
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\arace\Downloads\Datos_actividad_2.csv')
```

```python
df.info()
```

```python
df['Ganancia'] = df['Ingresos'] - df['Gastos']

df['Margen porcentual'] = (df['Ganancia'] / df['Ingresos']) * 100

print(df[['Sucursal', 'Ingresos', 'Gastos', 'Ganancia', 'Margen porcentual']])
```

```python
df
```

```python
df[['Sucursal', 'Ingresos', 'Gastos' ]].min()
df[['Sucursal', 'Ingresos', 'Gastos' ]].max()
```

```python
df[['Ingresos', 'Gastos']].var()
df[['Ingresos', 'Gastos']].std()
df[['Ingresos', 'Gastos', 'Ganancia']].std()
sucursal = df.loc[df['Gastos'] == df['Gastos'].max()]
```

```python
print("Desviación estándar:")
print(df[['Ingresos', 'Gastos', 'Ganancia']].std())
```

```python
df['Fecha'] = pd.to_datetime(df['Fecha'])

resumen_mensual = df.groupby(df['Fecha'].dt.month)[['Ingresos', 'Gastos']].sum()

print("Evolución mensual de ingresos y gastos:")
print(resumen_mensual)
```

```python
top5 = df.groupby('Sucursal')[['Ingresos', 'Gastos']].sum().nlargest(5, 'Ingresos')
```

```python
resumen_mensual['top 5'] = resumen_mensual.index.isin(top5.index)
resumen_mensual
```

```python
resumen_mensual['Estrategia'] = ''

resumen_mensual.loc[
    resumen_mensual['Gastos'] < resumen_mensual['Ingresos'],
    'Estrategia'
] = 'Reducir gastos'

resumen_mensual.loc[
    resumen_mensual['Ingresos'] < resumen_mensual['Gastos'],
    'Estrategia'
] = 'Mantener gastos'

resumen_mensual
```

```python
resumen_mensual.to_excel('estrategias.xlsx')
```
